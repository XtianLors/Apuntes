```python

#Add input to functions:
def askForInput(func):
    '''
    Decorator asks for user input at terminal:
    '''
    @wraps(func)
    def wrapper():
        yourFile = input('Name of the file: ')
        tabName = input('Name the table to create: ')
        return func(yourFile, tabName)
    return wrapper
    
```
