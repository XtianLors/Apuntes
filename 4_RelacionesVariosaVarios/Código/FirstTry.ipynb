```python
import sqlite3

fileName = "SecondTry.sqlite"

conn = sqlite3.connect(fileName)
cur = conn.cursor()

# Make some fresh tables using executescript()
cur.executescript('''
DROP TABLE IF EXISTS Artist;
DROP TABLE IF EXISTS Genre;
DROP TABLE IF EXISTS Album;
DROP TABLE IF EXISTS Track;

CREATE TABLE Artist (
    id  INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT UNIQUE,
    name    TEXT UNIQUE
);

CREATE TABLE Genre (
    id  INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT UNIQUE,
    name    TEXT UNIQUE
);

CREATE TABLE Album (
    id  INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT UNIQUE,
    artist_id  INTEGER,
    title   TEXT UNIQUE
);

CREATE TABLE Track (
    id  INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT UNIQUE,
    title TEXT  UNIQUE,
    album_id  INTEGER,
    genre_id  INTEGER,
    len INTEGER, rating INTEGER, count INTEGER
);
''')
cur.close()
conn.close()

conn = sqlite3.connect(fileName)
conn.row_factory = sqlite3.Row
cur = conn.cursor()
handle = open(r'.\tracks\tracks\tracks.csv')
#Another One Bites The Dust,Queen,Greatest Hits,55,100,217103,Rock
#   0                          1      2           3  4   5     6
for line in handle:
    line = line.strip();
    pieces = line.split(',')
    if len(pieces) < 7 : continue

    artist = pieces[1]
    track = pieces[0]
    genre = pieces[6]
    album = pieces[2]
    length = pieces[5]
    rating = pieces[4]
    count = pieces[3]
    #print(name, artist, album, count, rating, length, genre)

    #Fill the Artist table
    artistInsert = cur.execute('SELECT name FROM Artist WHERE name = (?) ', (artist,))
    row = artistInsert.fetchone()
    if row is None:
        cur.execute('INSERT OR IGNORE INTO Artist (name) VALUES (?)', (artist,))
    else:
        pass
        
    #Fill the Genre table
    genreInsert = cur.execute('SELECT name FROM Genre WHERE name = (?) ', (genre,))
    row = genreInsert.fetchone()
    if row is None:
        cur.execute('INSERT OR IGNORE INTO Genre (name) VALUES (?)', (genre,))
    else:
        pass
        
    #Fill the Album table with a foreign key comming from Artist table
    albumInsert = cur.execute('SELECT title FROM Album WHERE title = (?) ', (album,))
    row = albumInsert.fetchone()
    if row is None:
        ret = cur.execute('SELECT id FROM Artist WHERE name = (?) ', (artist,))
        artist_id = ret.fetchone()
        artist_id = artist_id[0]
        cur.execute('INSERT OR IGNORE INTO Album (artist_id, title) VALUES (?, ?)', (artist_id, album))
    else:
        pass

    #Fill the Track table with a foreign key comming from Artist and Genre tables.
    #Track(id, title, album_id, genre_id, len, rating, count)
    #Track(?, ?, ?, ?, ?, ?, ?)
    trackInsert = cur.execute('SELECT title FROM Track WHERE title = (?) ', (track,))
    row = trackInsert.fetchone()
    if row is None:
        ret = cur.execute('SELECT id FROM Genre WHERE name = (?) ', (genre,))
        genre_id = ret.fetchone()
        genre_id = genre_id[0]
        ret = cur.execute('SELECT id FROM Album WHERE title = (?) ', (album,))
        album_id = ret.fetchone()
        album_id = album_id[0]
        cur.execute('INSERT OR IGNORE INTO Track (title, album_id, genre_id, len, rating, count) VALUES (?, ?, ?, ?, ?, ?)', (track, album_id, genre_id, length, rating, count))
    else:
        pass

conn.commit()
cur.close()
conn.close()

```
