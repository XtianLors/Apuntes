## First queries, Joining data together.

- Often when we want some data it comes from a number of tables linked by these foreign keys.


```python
print('SELECT Album.title, Artist.name FROM Album JOIN Artist ON Album.artist_id = Artist.id')
print('SELECT ALBUM.title, Album.artist, Artist.id, Artist.name FROM Album JOIN Artist ON Album.artist_id = Artist.id')
print('SELECT Track.title, Genre.name FROM Track JOIN Genre ON Track.genre_id = Genre.id')
```
