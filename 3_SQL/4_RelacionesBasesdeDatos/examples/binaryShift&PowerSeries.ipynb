```python
# Left shift (<<) multiplies by 2^n
a = 14 # binary: 1110
print(a << 2) # binary: 111000 -> decimal: 56

# Right shift (>>) divides by 2^n (floor division)
b = 14 # binary: 1110
print(b >> 2) # binary: 11 -> decimal: 3
```

```python
n = 3

result = a << 3 # equivalent to a * (2 ** n)
result
```
